```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
Houseprice_data = pd.read_csv("/content/BostonHousing.csv")
```

```python
Houseprice_data.head()
```

```python
Houseprice_data.shape
```

```python
Houseprice_data.isnull().sum()
```

```python
Houseprice_data.describe()
```

```python
correlation = Houseprice_data.corr()
```

```python
# making heat map
plt.figure(figsize=(10,10))

sns.heatmap(correlation, cbar= True, square=True, fmt='.1f', annot=True,annot_kws={'size': 8},cmap='Reds')
```

```python
x = Houseprice_data.drop(['price'],axis = 1)
y = Houseprice_data['price']
```

```python
print(x,y)
```

```python
# traing and test data

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size= 0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
# XgBoost regrssor

model = XGBRegressor()
```

```python
model.fit(x_train,y_train)
```

```python
# evaluation


training_data_pridiction = model.predict(x_train)
```

```python
print(training_data_pridiction)
```

```python
# r squired error

score_1 = metrics.r2_score(y_train,training_data_pridiction)
```

```python
# mean absolute error

score_2 = metrics.mean_absolute_error(y_train,training_data_pridiction)
```

```python
print("r squired error = ",score_1)
print('mean absolute error = ',score_2)
```

```python
testing_data_pridiction = model.predict(x_test)
print(training_data_pridiction)
```

```python
score_3 = metrics.r2_score(y_test,testing_data_pridiction)
score_4 = metrics.mean_absolute_error(y_test,testing_data_pridiction)
print("r squired error for tast data = ",score_3)
print('mean absolute error for test data = ',score_4)
```

```python
# visualizing the data

plt.scatter(y_train,training_data_pridiction)
plt.xlabel("Actual Prices")
plt.ylabel("pridicted Prices")
plt.title("Actual house price vs predicted price")
plt.Colorizer("Blues_r")
plt.show()
```

```python
plt.scatter(y_test,testing_data_pridiction)
plt.xlabel("Actual Prices")
plt.ylabel("pridicted Prices")
plt.title("Actual house price vs predicted price")
plt.Colorizer('Greys')
plt.show()
```
